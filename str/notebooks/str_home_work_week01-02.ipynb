## Partie 4 : Simulation Python

Modifiez les paramètres `C`, `T` et `duree_simulation` dans la cellule suivante. Observez les échéances respectées ou violées.

Testez au minimum les cas suivants :
- `C = 2`, `T = 5` ;
- `C = 5`, `T = 5` ;
- `C = 6`, `T = 5`.

Expliquez dans la cellule Markdown suivante à partir de quelle condition l'ordonnancement devient impossible.

```python
import matplotlib.pyplot as plt

# Paramètres à modifier
C = 6
T = 5
duree_simulation = 30

activations = list(range(0, duree_simulation, T))
fin_execution = [min(activation + C, duree_simulation) for activation in activations]

fig, ax = plt.subplots(figsize=(11, 2.8))
for numero, (activation, fin) in enumerate(zip(activations, fin_execution), 1):
    ax.barh(0.25, fin - activation, left=activation, height=0.2, color="#2f80ed")
    ax.text((activation + fin) / 2, 0.25, f"Instance {numero}", ha="center", va="center", color="white", fontsize=8)
    ax.annotate("", xy=(activation + T, 0.02), xytext=(activation + T, 0.35), arrowprops={"arrowstyle": "<->", "color": "#d64545", "linestyle": "--"})

ax.spines["bottom"].set_position(("data", 0))
ax.spines["top"].set_visible(False)
ax.set_xlim(0, duree_simulation)
ax.set_ylim(-0.1, 0.7)
ax.set_yticks([0.25])
ax.set_yticklabels(["Processeur"])
ax.set_xticks(range(0, duree_simulation + 1, T))
ax.set_xlabel("Temps")
ax.set_title(f"Tâche périodique : C={C}, T={T}")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()
```

D'après les simulations de la partie 4, lorsque C = 2 et T = 5, l'ordonnancement est possible car le temps d'exécution est inférieur à la période. Lorsque C = 5, le processeur est utilisé à 100 % et il n'y a plus de marge disponible. Enfin, lorsque C = 6 et T = 5, l'ordonnancement devient impossible car le temps d'exécution dépasse la période. La tâche ne peut donc pas terminer avant l'arrivée de la prochaine instance, ce qui entraîne un non-respect des échéances.

### Interprétation de la simulation

1. Pour quelle relation entre `C` et `T` chaque instance respecte-t-elle son échéance ?
2. Que se passe-t-il lorsque `C > T` ?
3. Pourquoi le cas `C = T` laisse-t-il une marge nulle au processeur ?

**Réponse :**

................................................................................
................................................................................

1-Chaque instance respecte son échéance si C<=T>. Si C<T, elle termine avant l'échéance et il reste une marge de T-C. Si C=T, elle termine exactement à l'échéance

2-Lorsque C>T, une instance n'a pas terminé au moment où la suivante est activée. Le retard s'accumule et les échéances sont violées : l'ordonnancement de cette tâche seule est impossible à long terme sur ce processeur

3-Lorsque C=T, toute la capacité du processeur est utilisée par cette tâche : U=C/T=1=100%. Il ne reste aucun temps libre pour les autres tâches, les interruptions ou les imprévus

La conception d'un système temps réel doit prendre en compte simultanément la justesse des résultats et les délais dans lesquels ils doivent être fournis.
